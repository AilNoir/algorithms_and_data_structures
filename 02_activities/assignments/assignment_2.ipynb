## Problem Two: Path to Leaves
**Description**  
Given the `root` of a binary tree, return all root to leaf paths in any order.

**Examples**

* Example 1:

    ![](./images/q1_ex1.png)

    Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?*

    Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

* Example 2:

    ![](./images/q1_ex3.png)

    Input: `root = [10, 9, 7, 8]`

    Output: [[10, 7], [10, 9, 8]]

**Problem 2 Starter Code**

You may assume that `root` is a binary tree that has been built, i.e., you can access a node/leaf by calling `root.left.right.left.right...`, so you don't have to provide a tree-building function to convert an input list to a binary tree as an object of the `TreeNode` class.

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right
from collections import deque
from typing import List
def bt_path(root: TreeNode) -> List[List[int]]:
  if root is None:
    return []
  result = []
  queue = deque([(root, [root.val])])
  while queue:
    node, path = queue.popleft()
    if node.left is None and node.right is None:
      result.append(path)
    if node.left:
      queue.append((node.left, path + [node.left.val]))
    if node.right:
      queue.append((node.right, path + [node.right.val]))
```

from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    new_list = [];
    zero_count = 0;
    
    for num in nums:
        if num != 0:
            new_list.append(num)
        else:
            zero_count += 1

    for i in range(zero_count):
        new_list.append(0);

    return new_list

5. Explain why their solution works in your own words.


The code successfully solves the problem by using a simple two-step process:
1. Collect non-zeros first: It loops through the list from left to right. Every time it sees a non-zero number, it immediately adds it to a new list. Since it processes them in order, the relative order of these numbers is perfectly kept.
2. Count and pad zeros: Instead of moving zeros immediately, it simply counts them. At the end, it appends that exact number of zeros to the back of the new list.


6. Explain the time and space complexity of their solution in your own words and identify if they found the correct time and space complexities.


Time Complexity: \(O(N)\)
The code loops through the list of size \(N\) twice. The first loop goes through every element to filter out non-zero numbers (\(O(N)\)). The second loop appends the remaining zeros, taking at most \(O(N)\) steps. Since \(O(N) + O(N) = O(2N)\), it simplifies to \(O(N)\) (Linear Time).
Space Complexity: \(O(N)\)
The code creates a completely new list (new_list) to store the results. In the worst-case scenario (e.g., if there are no zeros), this new list will grow to the exact same size as the input list, occupying \(O(N)\) extra memory (Linear Space).

7. Suggest if any improvements can be made to your partner's solution.
